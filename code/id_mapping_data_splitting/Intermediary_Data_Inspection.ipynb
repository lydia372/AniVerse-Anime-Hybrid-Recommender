```python
# anime master features

import numpy as np
import pandas as pd

from google.colab import drive
drive.mount('/content/drive')

anime_features = np.load("/content/drive/MyDrive/MASTER_ANIME_TOWER_FEATURES_1BASED.npy", allow_pickle=True)

print("=" * 60)
print("ANIME FEATURE MATRIX")
print("=" * 60)
print(f"Shape          : {anime_features.shape}")
print(f"  → rows       : {anime_features.shape[0]}  (row 0 = padding, rows 1+ = anime)")
print(f"  → columns    : {anime_features.shape[1]}  (total features per anime)")
print(f"Dtype          : {anime_features.dtype}")
print(f"\nPadding row (row 0) — should be all zeros:")
print(f"  All zeros?   : {np.all(anime_features[0] == 0)}")
print(f"  Row 0 values : {anime_features[0][:10]} ...")

print(f"\nSample anime row (row 1):")
print(f"  Values       : {anime_features[1][:10]} ...")
print(f"\nValue range across entire matrix:")
print(f"  Min          : {np.nanmin(anime_features):.4f}")
print(f"  Max          : {np.nanmax(anime_features):.4f}")
print(f"  Mean         : {np.nanmean(anime_features):.4f}")
print(f"  NaN count    : {np.sum(np.isnan(anime_features.astype(float)))}")

```
